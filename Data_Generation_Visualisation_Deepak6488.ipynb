```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ensure plots render inline in Jupyter
%matplotlib inline

# 1. Create a Pandas DataFrame with four columns (f1, f2, f3, f4) populated with random numerical values
np.random.seed(42)  # Optional: ensures reproducible random numbers
data = np.random.randint(10, 100, size=(20, 4))
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])

print("Initial DataFrame (f1 to f4):")
display(df.head())

# 2. Generate a scatter plot illustrating the relationship between column f2 and column f3
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], color='teal', edgecolors='black', s=70, alpha=0.85)
plt.title('Scatter Plot: Relationship between f2 and f3', fontsize=13, fontweight='bold')
plt.xlabel('Column f2 Values', fontsize=11)
plt.ylabel('Column f3 Values', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 3. Compute the row-wise arithmetic mean across f1, f2, f3, and f4 and store in 'f5'
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

print("Complete DataFrame (f1 through f5):")
display(df)

# 4. Export the complete DataFrame to a Microsoft Excel file (.xlsx)
output_file = 'Exercise_1_EDA_Report.xlsx'
df.to_excel(output_file, index=False, sheet_name='EDA_Data')

print(f"DataFrame successfully exported to '{output_file}'!")
```
